```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
df = pd.read_csv(
    "../data/raw_cbc.csv",
    skiprows=[1]
)
print(df.head())
```

```python
print("Dataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nFirst Five Records:")
display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df = df.dropna(how='all')

df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
df = df.drop(
    columns=['S. No.'],
    errors='ignore'
)
```

```python
print("Before removing blank rows:", df.shape)

df = df.dropna(how='all')

print("After removing blank rows:", df.shape)
```

```python
print(df.isnull().sum())
```

```python
df = df.rename(columns={
    'S. No.': 'Serial_No',
    'Sex': 'Gender',
    'TLC': 'WBC',
    'PLT /mm3': 'PLT'
})

print(df.columns.tolist())
```

```python
print("Duplicate records:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df['Leukopenia_Risk'] = (df['WBC'] < 4.0).astype(int)

print(df['Leukopenia_Risk'].value_counts())
```

```python
print(df['Leukopenia_Risk'].value_counts(normalize=True) * 100)
```

```python
print("WBC values <= 0:", (df['WBC'] <= 0).sum())
print("HGB values <= 0:", (df['HGB'] <= 0).sum())

print(df.describe())
```

```python
print(df.isnull().sum())
print(df.dtypes)
```

```python
df.to_csv('../data/cleaned_cbc.csv', index=False)

print("Cleaned dataset saved successfully!")
```
